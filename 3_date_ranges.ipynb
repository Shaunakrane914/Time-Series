```python
import pandas as pd
```

```python
#Generate daily dates

dates = pd.date_range(start ="2026-01-01",
                      periods = 7,
                      freq="D" 
                      #Calender Day Everyday(Monday to Sunday)        
                              
                              )

print(dates)
```

```python
pd.date_range(

    start="2024-01-01",
    periods = 10,
    freq="B" #Business Day Monday to Friday only(weekends excluded)
)
```

```python
#Generate Dates Between Two dates

dates = pd.date_range(
      start = "2026-01-01",
      end = "2026-01-10"

)
dates
```

```python
#Monthly Dates
pd.date_range(
    start="2026-01-01",
    periods=12,
    freq="M"
)
```

```python
#Monthly Dates
pd.date_range(
    start="2026-01-01",
    periods =12,
    freq="MS"
)
```

```python
#Weekly dates
pd.date_range(
   start="2026-01-01",
   periods=12,
   freq="H"
)
```

```python
#Hourly Data
pd.date_range(
    start = "2026-01-01 09:00",
   periods=12,
  freq="H"

)
```

```python
#Every 15 Minutes
pd.date_range(
    start = "2026-01-01 09:00",
   periods=20,
   freq ="15min"

)
```

```python
#Create a Time Series
dates = pd.date_range(
start = "2026-01-01",
periods=7

)

sales = [120,135,150,145,160,170,180]

ts = pd.Series(sales, index = dates)
print(ts)
```
