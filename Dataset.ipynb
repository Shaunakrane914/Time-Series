```python
import pandas as pd

dates = pd.date_range(start="2026-01-01", periods=10, freq="D")
sales_data = pd.DataFrame({"date": dates, "sales": [120, 135, None, 142, 150, 147, 160, None, 168, 180]})
sales_data

```

```python
data_fill = sales_data.copy()
data_fill["sales"] = data_fill["sales"].ffill()
print(data_fill)
```

```python
data_bfill = sales_data.copy()
data_bfill["sales"] = data_bfill["sales"].bfill()
print(data_bfill)
```

```python
data_inter = sales_data.copy()
data_inter["sales"] = data_inter["sales"].interpolate()
print(data_inter)
```

```python
import pandas as pd

dates = pd.date_range(start="2026-01-01", periods=5, freq="D")
sales_data1 = pd.DataFrame({"date": dates, "sales": [1200, 1300, 1400, 1500, None]})
sales_data1
```
