```python
from statsmodels.tsa.seasonal import seasonal_decompose 
import pandas as pd
df = pd.read_csv("./data/Time Series Composition/tourist_arrivals_5_years.csv")
df.head()
```

```python
df.info()
```

```python
import datetime
```

```python
df["Date"] = pd.to_datetime(
  df["Date"]
)
```

```python
df = df.set_index("Date")
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12,4))
plt.plot(df.index,df["Tourist_Arrivals"], linewidth = 2)

plt.title("Tourist Arrivals(2021-2025)")
plt.xlabel("Year")
plt.ylabel("Number of Tourist")
plt.grid(True, linestyle=":")

plt.show()
```

```python
result = seasonal_decompose(
df["Tourist_Arrivals"],
model = "multiplcative",
period = 12
)
```
